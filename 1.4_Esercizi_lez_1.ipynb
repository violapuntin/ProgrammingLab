### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
numeri = [1, 2, 3, 4, 5]
quadrati = []

for n in numeri:
    quadrati.append(n**2)

print(quadrati) # Output: [1, 4, 9, 16, 25]
```

```python
numeri = [1, 2, 3, 4, 5]
quadrati = [n**2 for n in numeri]
print(quadrati) # Output: [1, 4, 9, 16, 25]
```

```python
numeri = [10, 15, 20, 25, 30]
pari = []

for n in numeri:
    if n % 2 == 0:
        pari.append(n)

print(pari) # Output: [10, 20, 30]
```

```python
pari = [n for n in numeri if n % 2 == 0]
```

```python
nomi = ["alice", "bob", "carla"]
nomi_maiuscoli = []

for nome in nomi:
    nomi_maiuscoli.append(nome.upper())

print(nomi_maiuscoli) # Output: ['ALICE', 'BOB', 'CARLA']
```

```python
nomi_maiuscoli = [nome.upper() for nome in nomi]
```

### **2. Vettore di Numeri Primi**

1. **Crea un vettore** contenente tutti i numeri primi compresi tra **0 e 10**  
   _(Puoi scriverli direttamente nell'array)._

2. **Conta quanti numeri ci sono nel vettore** utilizzando la funzione `len()`.  **Ottieni lo stesso numero** accedendo all'attributo `.size` del vettore.

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
import numpy as np

# Creazione del vettore manuale
vettore_primi = np.array([2, 3, 5, 7])
print("Vettore:", vettore_primi)

conteggio_len = len(vettore_primi)
conteggio_size = vettore_primi.size

print(f"Lunghezza con len(): {conteggio_len}")
print(f"Grandezza con .size: {conteggio_size}")

print("Il tipo di dato è:", vettore_primi.dtype)

# List comprehension con controllo numeri primi
# Un numero n è primo se non è divisibile per nessun i tra 2 e la sua radice quadrata
vettore_comp = np.array([n for n in range(2, 11) if all(n % i != 0 for i in range(2, int(n**0.5) + 1))])

print("Vettore da list comprehension:", vettore_comp)
```

### **3.  Operazioni su array**
Crea un array **a**, 1D (senza digitarlo esplicitamente), e genera due nuovi array **b**, contenente un sottoinsieme a piacere e **c** che è il reverse dell'array **a** (esempio [1,2,3] diventa [3,2,1]) .  Dividi l'array **a** per l'array **c**. Fai la stessa cosa per una lista.

```python
import numpy as np

# Creiamo l'array 'a' (es. numeri da 1 a 5) senza digitarli
a = np.arange(1, 6) 

# Creiamo 'b' prendendo una sottostringa (slicing)
# Prendo i primi tre elementi [1, 2, 3]
b = a[0:3]

# Creiamo 'c' che è il reverse di 'a'
c = a[::-1]

# Divisione tra array (Vettorizzata)
risultato_numpy = a / c

print(f"Array a: {a}")
print(f"Array b (sottostringa): {b}")
print(f"Array c (reverse): {c}")
print(f"Divisione a/c: {risultato_numpy}")
```

```python
# Creiamo le liste partendo da quelle NumPy o con range()
lista_a = list(range(1, 6))
lista_b = lista_a[0:3]
lista_c = lista_a[::-1]

# La divisione DIRETTA darebbe errore: lista_a / lista_c  <-- NON FUNZIONA

# Dobbiamo usare una list comprehension
risultato_lista = [val_a / val_c for val_a, val_c in zip(lista_a, lista_c)]

print(f"Lista a: {lista_a}")
print(f"Lista c: {lista_c}")
print(f"Divisione con list comprehension: {risultato_lista}")
```
